```python

# Import required libraries for loading and running the model
import torch
import transformers
from transformers import AutoModelForCausalLM, AutoTokenizer

# Model name (pretrained chat model from Hugging Face)
model_id = "mistralai/Mistral-7B-Instruct-v0.1"

# Settings to make the model smaller and faster using 4-bit loading
bnb_config = transformers.BitsAndBytesConfig(
    load_in_4bit=True,  # reduce memory usage
    bnb_4bit_use_double_quant=True,  # better compression
    bnb_4bit_quant_type="nf4",  # quantization method
    bnb_4bit_compute_dtype=torch.float16  # faster computation
)

# Load the language model
model = AutoModelForCausalLM.from_pretrained(
    model_id,
    device_map="auto",  # automatically use CPU or GPU if available
    torch_dtype=torch.float16,  # use half precision for speed
    trust_remote_code=True  # allow custom model code
)

# Load tokenizer (converts text ↔ numbers so model can understand input/output)
tokenizer = AutoTokenizer.from_pretrained(model_id)
```

Create AI Text Generator

```python
# Create a text generator using the model and tokenizer
# It makes the model write answers from a prompt
pipe = pipeline(
    "text-generation",   # tells model to generate text
    model=model,         # the AI model
    tokenizer=tokenizer, # helps convert text to numbers and back
    max_new_tokens=200   # limits how long the answer can be
)

# Make the model usable in LangChain
# This lets us use it in RAG
hf = HuggingFacePipeline(pipeline=pipe)
```

**RAG Pipeline**

```python
from langchain_core.prompts import PromptTemplate
from langchain_core.runnables import RunnablePassthrough
from langchain_community.llms import HuggingFacePipeline
from langchain_community.vectorstores import FAISS
from langchain_core.prompts import PromptTemplate
from langchain_core.runnables import RunnablePassthrough, RunnableLambda
```

```python
##Prompt
# This is the instruction format given to the AI model

# The model is told to:
# - ONLY use the provided context (no outside knowledge)
# - read the context
# - then answer the question
# The [INST] ... [/INST] format is required by Mistral models to structure instructions properly
prompt_template = """
[INST]
Answer ONLY using the context below.

Context:
{context}

Question:
{question}
[/INST]
"""
```

```python
# Create a prompt object that defines what inputs the AI needs
# It tells the model we will give:
# - context (extra information from documents)
# - question (user's question)

prompt = PromptTemplate(
    input_variables=["context", "question"],  # required inputs for the prompt
    template=prompt_template,                 # the format of the instruction
)
```

Build Input Function for RAG Pipeline (Retrieval + Context Preparation)

```python
# This function prepares input for the RAG system

# It takes the user question and:
# 1. Finds similar documents from the database
# 2. Combines them into a single context
# 3. Returns both context and question for the AI

def build_inputs(inputs):

    # If input is just a text string, treat it as the question
    if isinstance(inputs, str):
        question = inputs
    else:
        # If input is a dictionary, extract the question
        question = inputs.get("question")

    # Search for similar documents in FAISS database
    docs = db.as_retriever().invoke(question)

    # Join all found documents into one text (context)
    context = "\n".join([d.page_content for d in docs])

    # Return both context and question for the model
    return {
        "context": context,
        "question": question
    }

```

```python
# RAG CHAIN
# This is the RAG system (chat with documents)

# It works in 3 steps:
# 1. Find related documents
# 2. Put them into a question format
# 3. Ask the AI model to answer

rag_chain = (
    RunnableLambda(build_inputs)  # gets related text from database
    | prompt                       # prepares the question format
    | hf                           # AI writes the final answer
)
```

```python
# This function cleans the AI output (final answer)

# It makes sure the output is always readable text


def parse_answer(output):
    if isinstance(output, str):
        # Strip everything up to and including [/INST]
        if "[/INST]" in output:
            return output.split("[/INST]")[-1].strip()
        return output.strip()
    if isinstance(output, dict):
        text = output.get("text", str(output))
        if "[/INST]" in text:
            return text.split("[/INST]")[-1].strip()
        return text.strip()
    return str(output)

```

```python
# This sends a question to the RAG system

# Step 1: user question is sent
# Step 2: system finds relevant documents
# Step 3: AI generates an answer

result = rag_chain.invoke({
    "question": "Tell me about David's Cookies Butter Pecan Meltaways"
})

# Clean and print the final answer
print(parse_answer(result))
```
